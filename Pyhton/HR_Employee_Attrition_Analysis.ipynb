# DEPARTMENT-WISE ATTRATION RATE


```python
dept_summary = df.groupby("Department").agg(
    total_employee = ("Attrition","count"),
    employee_left = ("Attrition",lambda x: (x== "Yes").sum())
)

dept_summary["Attrition_rate"] = (
    dept_summary["employee_left"]/dept_summary["total_employee"] * 100).round(2)

dept_summary
```

```python
dept_summary.sort_values("Attrition_rate", ascending=False)
```

```python
dept_summary = dept_summary.sort_values("Attrition_rate", ascending=False)

dept_summary["Attrition_rate"].plot(kind="bar")

plt.title("Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

### Department-wise Attrition – Key Insight

Sales has the highest employee attrition rate at 20.63%, followed by Human Resources at 19.05%, while Research & Development has the lowest attrition rate at 13.84%.

Although R&D has the highest number of employees leaving, its larger workforce results in a lower attrition rate. This suggests that Sales may require greater attention from HR when investigating the drivers of employee turnover.

## Analysis 2: Overtime vs Attrition


Are employees who work overtime more likely to leave the company?

```python
overtime_attrition = pd.crosstab(
    df["OverTime"],
    df["Attrition"]
)

overtime_attrition
```

```python
overtime_rate = pd.crosstab(
    df["OverTime"],
    df["Attrition"],
    normalize = "index"
)* 100

overtime_rate.round(2)
```

```python
overtime_rate["Yes"].plot(kind="bar")

plt.title("Attrition Rate by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

### Overtime vs Attrition – Key Insight

Employees working overtime have a significantly higher attrition rate (30.53%) compared with employees who do not work overtime (10.44%). This indicates a strong association between overtime and employee attrition.

However, this analysis alone does not establish that overtime directly causes employees to leave. Further analysis of factors such as job satisfaction, work-life balance, income, and job role can help identify the underlying drivers.

## Final Visualization

### 1. Overall Employee Attrition


```python
attrition_count = df["Attrition"].value_counts()
attrition_percentage = df["Attrition"].value_counts(normalize=True) * 100

plt.figure(figsize=(6, 4))

bars = plt.bar(
    attrition_count.index,
    attrition_count.values
)

plt.title("Overall Employee Attrition")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

for bar, count, percentage in zip(
    bars,
    attrition_count.values,
    attrition_percentage.values
):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f"{count}\n({percentage:.1f}%)",
        ha="center",
        va="bottom"
    )

plt.show()
```

Key Insight:
The dataset contains 1,470 employees, of which 237 employees left the organization, resulting in an overall attrition rate of 16.12%.
