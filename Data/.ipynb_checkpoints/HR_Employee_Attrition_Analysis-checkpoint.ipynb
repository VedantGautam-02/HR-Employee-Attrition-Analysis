## Overall attration rate

(KPI 1)Calculating total employees for kpi


```python
total_employees = len(df)
print('total empoyees:', total_employees)
```

(KPI 2) for calculating how many employees are left

```python
employees_left = (df["Attrition"] == "Yes").sum()
print("total employees that have left:" ,employees_left)
```

(KPI 3) calculating the number of employee stayed 


```python
employees_stayed = (df["Attrition"] == "No").sum()

print(" total employees Stayed:", employees_stayed)
```

(KPI 4) Calculating attrition rate

```python
attrition_rate = (employees_left / total_employees) * 100
print(f"Attrition Rate: {attrition_rate:.2f}%")
```
